```python
import requests

latitude = 19.9000   #19.9105          # 19.9000,
longitude = 99.8333   #99.8406         # 99.8333

url = (
    "https://api.open-meteo.com/v1/forecast"
    f"?latitude={latitude}"
    f"&longitude={longitude}"
    "&current=temperature_2m,relative_humidity_2m"
)

response = requests.get(url)

data = response.json()

temperature = data["current"]["temperature_2m"]
humidity = data["current"]["relative_humidity_2m"]

print("Temperature =", temperature, "°C")
print("Humidity =", humidity, "%")
```

```python
if temperature > 35:
    print("High Temperature")

elif temperature > 30:
    print("Warm Weather")

else:
    print("Normal")
```

```python
import requests
import pandas as pd

url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=19.9105"
    "&longitude=99.8406"
    "&hourly=temperature_2m"
)

data = requests.get(url).json()

time = data["hourly"]["time"]

temp = data["hourly"]["temperature_2m"]

df = pd.DataFrame({
    "Time": time,
    "Temperature": temp
})

print(df.head())

print(df.describe())
```

```python
import requests
import numpy as np

url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=19.9105"
    "&longitude=99.8406"
    "&hourly=temperature_2m"
)

data = requests.get(url).json()

temp = data["hourly"]["temperature_2m"]

# Data
print("Raw Data")
print(temp[:10])

# Information
avg = np.mean(temp)

max_temp = np.max(temp)

print("\nAverage =", avg)

print("Maximum =", max_temp)

# Knowledge

if avg > 30:

    knowledge = "Weather is generally hot"

else:

    knowledge = "Weather is normal"

print("\nKnowledge =", knowledge)

# Decision

if avg > 35:

    decision = "Turn on air conditioning"

elif avg > 30:

    decision = "Turn on ventilation"

else:

    decision = "Normal"

print("\nDecision =", decision)
```

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=19.9105"
    "&longitude=99.8406"
    "&hourly=temperature_2m"
)

data = requests.get(url).json()

time = data["hourly"]["time"]

temp = data["hourly"]["temperature_2m"]

df = pd.DataFrame({
    "Time": time,
    "Temperature": temp
})

# แสดงเฉพาะ 24 ชั่วโมงแรก

df = df.head(24)

plt.figure(figsize=(12,5))

plt.plot(df["Time"], df["Temperature"])

plt.xticks(rotation=45)

plt.xlabel("Time")

plt.ylabel("Temperature (°C)")

plt.title("Temperature Forecast (24 Hours)")

plt.tight_layout()

plt.show()
```
